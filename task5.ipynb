```python
# 5.a
from math import comb
# Функция подсчёта вероятности биномиального распределения
def bin(m,n,p):
    c_n_m = comb(n,m)
    p_m = p**m
    q_n_m = (1-p)**(n-m)
    return c_n_m*p_m*q_n_m

n = 1000
p = 0.005

bin_p_x_3 = bin(3, n, p)
print(bin_p_x_3)
bin_p_x_2 = bin(2, n, p)
bin_p_x_1 = bin(1, n, p)
bin_p_x_0 = bin(0, n, p)
bin_p_x_geq_3 = 1-bin_p_x_2-bin_p_x_1-bin_p_x_0
print(bin_p_x_geq_3)
```

```python
# 5.b
from math import factorial
import math

# Функция для расчёта вероятности распределения Пуассона
def poiss(lamb, m):
    l_m = lamb**m
    m_fact = factorial(m)
    e_lamb = math.e**(-lamb)
    return l_m/m_fact * e_lamb

lamb = n*p

poiss_p_x_3 = poiss(lamb, 3)
print(poiss_p_x_3)

poiss_p_x_2 = poiss(lamb, 2)
poiss_p_x_1 = poiss(lamb, 1)
poiss_p_x_0 = poiss(lamb, 0)
poiss_p_x_geq_3 = 1 - poiss_p_x_2 - poiss_p_x_1 - poiss_p_x_0
print(poiss_p_x_geq_3)
```

```python
# 5.c compare

Delta_p_x_3 = abs(poiss_p_x_3 - bin_p_x_3)
delta_p_x_3 = Delta_p_x_3 / bin_p_x_3


Delta_p_x_geq_3 = abs(poiss_p_x_geq_3 - bin_p_x_geq_3)
delta_p_x_geq_3 = Delta_p_x_geq_3 / bin_p_x_geq_3

print(f"Абсолютная погрешность P(X=3): {Delta_p_x_3}")
print(f"Относительная погрешность P(X=3): {delta_p_x_3}")
print()
print(f"Абсолютная погрешность P(X>=3): {Delta_p_x_geq_3}")
print(f"Относительная погрешность P(X>=3): {delta_p_x_geq_3}")

```

```python
# 5.d
import numpy as np

# Функция моделирования
rng = np.random.default_rng()

# Функция подсчёта мат ожидания выборка
# Выборочное среднее
def E(array: list) -> float:
    return sum(array)/len(array)

# Функия подсчёта дисперсии по выборке
def variance(array: list) -> float:
    var = 0
    math_expect = E(array)
    for i in range(len(array)):
        var += (array[i] - math_expect)**2
    return var/len(array)

def experiment(N: int, bin_p_x_3: float, bin_p_x_geq_3: float) -> None:
    poiss_N = rng.poisson(lamb, N)
    E_poiss_N = E(poiss_N)
    Var_poiss_N = variance(poiss_N)

    filtered_poiss_N_X_eq_3 = list(filter(lambda x: x==3, poiss_N))
    filtered_poiss_N_X_geq_3 = list(filter(lambda x: x>=3, poiss_N))

    probability_poiss_N_X_eq_3 = len(filtered_poiss_N_X_eq_3)/N
    probability_poiss_N_X_geq_3 = len(filtered_poiss_N_X_geq_3)/N

    print(f"Вероятность P(X=3) при N:{N} -", probability_poiss_N_X_eq_3)
    print(f"Вероятность P(X>=3) при N:{N} -", probability_poiss_N_X_geq_3)

    print(f"Математическое ожидание при {N} -", E_poiss_N)
    print(f"Дисперсия при {N} -", Var_poiss_N)

    Delta_p_x_3 = abs(probability_poiss_N_X_eq_3 - bin_p_x_3)
    delta_p_x_3 = Delta_p_x_3 / bin_p_x_3

    print(f"Абсолютная погрешность P(X=3): {Delta_p_x_3}")
    print(f"Относительная погрешность P(X=3): {delta_p_x_3}")

    Delta_p_x_geq_3 = abs(probability_poiss_N_X_geq_3 - bin_p_x_geq_3)
    delta_p_x_geq_3 = Delta_p_x_geq_3 / bin_p_x_geq_3

    print(f"Абсолютная погрешность P(X>=3): {Delta_p_x_geq_3}")
    print(f"Относительная погрешность P(X>=3): {delta_p_x_geq_3}")
    
```

```python
n = 1000
p = 0.005
bin_p_x_3 = bin(3, n, p)
bin_p_x_2 = bin(2, n, p)
bin_p_x_1 = bin(1, n, p)
bin_p_x_0 = bin(0, n, p)
bin_p_x_geq_3 = 1-bin_p_x_2-bin_p_x_1-bin_p_x_0

experiment(100, bin_p_x_3, bin_p_x_geq_3)
print()
experiment(1000, bin_p_x_3, bin_p_x_geq_3)
print()
experiment(10000, bin_p_x_3, bin_p_x_geq_3)
```
